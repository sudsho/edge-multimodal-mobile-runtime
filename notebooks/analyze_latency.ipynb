# Latency analysis

Plot the on-device p50 / p90 numbers from `benchmarks/results.md` and
compare against the sub-30 ms end-to-end target.

```python
import matplotlib.pyplot as plt
import numpy as np

devices = ["iPhone 14 Pro ANE", "iPhone 11 ANE", "Pixel 7 NNAPI", "Samsung S22 NNAPI", "Pixel 4a TFLite"]
wake_p50 = np.array([1.8, 3.1, 4.2, 4.7, 8.2])
spk_p50  = np.array([11.4, 18.7, 22.8, 24.6, np.nan])
```

```python
fig, ax = plt.subplots(figsize=(8, 4))
x = np.arange(len(devices))
w = 0.35
ax.bar(x - w/2, wake_p50, w, label="wake p50")
ax.bar(x + w/2, spk_p50,  w, label="speaker p50")
ax.set_xticks(x)
ax.set_xticklabels(devices, rotation=20, ha="right")
ax.set_ylabel("latency (ms)")
ax.axhline(30, ls="--", color="grey", label="30 ms budget")
ax.legend()
ax.set_title("on-device p50 latency")
plt.tight_layout()
plt.show()
```

```python
total = wake_p50 + np.where(np.isnan(spk_p50), 0, spk_p50)
for d, t in zip(devices, total):
    tag = "OK" if t <= 30 else "OVER"
    print(f"{d:20s}  end-to-end p50 {t:5.1f} ms  {tag}")
```

## takeaways

- Every ANE path is well under budget.
- Pixel 7 and S22 land inside 30 ms.
- Pixel 4a is CPU-only for the speaker path; that device gets the
  wake-word only build and skips the speaker verify.
